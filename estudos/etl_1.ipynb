```python
import pandas as pd
```

```python
df = pd.read_csv('../dataset/dataset_v2.csv',header=None, names=['timestamp', 'variavel', 'valor'])
```

```python
#variaveis do sistema pneumático (atuador) que escolhemos
variaveis_desejadas = ['Avancado_1S2', 'Recuado_1S1', 'V1_14', 'V1_12', 'Inicia', 'Para']
```

```python
df = df[df['variavel'].isin(variaveis_desejadas)]
df.sort_values(by=['timestamp'], inplace=True)
```

```python
#df['variavel'].value_counts()
```

```python
df_pivot = df.pivot_table(
    index='timestamp', 
    columns='variavel', 
    values='valor', 
    aggfunc='first'  # caso haja duplicatas, pega o primeiro valor
).reset_index()
```

```python
df_pivot.to_excel('pivot_dataset_2.xlsx', index=False)
print(f"✅ DataFrame exportado")
```
