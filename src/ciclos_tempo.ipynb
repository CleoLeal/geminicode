```python
import pandas as pd
```

```python

# Carregar dataset pivotado
df = pd.read_excel("pivot_dataset.xlsx")
```

```python
ciclos = 0
estado = "RECUADO"
ciclo_col = []

for i, row in df.iterrows():
    if estado == "RECUADO" and row["Avancado_1S2"] == 1:
        estado = "AVANCADO"
    elif estado == "AVANCADO" and row["Recuado_1S1"] == 1:
        estado = "RECUADO"
        ciclos += 1  # fechou um ciclo
    ciclo_col.append(ciclos)
```

```python
# Adiciona a coluna de ciclos ao DataFrame
df["Ciclo"] = ciclo_col

# Salvar em Excel
df.to_excel("dataset_com_ciclos.xlsx", index=False)

print("Arquivo 'dataset_com_ciclos.xlsx' gerado com sucesso!")
```

```python
# Garantir que o timestamp seja datetime
df["timestamp"] = pd.to_datetime(df["timestamp"])
# Calcular tempo por ciclo
tempos_ciclos = []
for ciclo, grupo in df.groupby("Ciclo"):
    if ciclo > 0:  # ignora linhas fora dos ciclos
        inicio = grupo["timestamp"].iloc[0]
        fim = grupo["timestamp"].iloc[-1]
        duracao = (fim - inicio).total_seconds()
        tempos_ciclos.append({"Ciclo": ciclo, "Tempo (s)": duracao})
```

```python
# Converter em DataFrame
df_tempos = pd.DataFrame(tempos_ciclos)
```

```python
# Salvar em Excel
df_tempos.to_excel("tempos_por_ciclo.xlsx", index=False)

print("✅ Arquivo 'tempos_por_ciclo.xlsx' gerado com sucesso!")
print(df_tempos)
```
